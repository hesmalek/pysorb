# pysorb Basic Usage

This notebook demonstrates lightweight pysorb workflows. Install with `pip install pysorb` or run from a local editable checkout with `pip install -e .`.

```python
import numpy as np

from pysorb.fit import fit_isotherm
from pysorb.mix import parse, iast
from pysorb.bt import run_simulation
from pysorb.crank import crank_uptake
from pysorb.zlc import linmodel
```

## Isotherm Fitting

```python
P = np.array([0.1, 0.3, 0.7, 1.2, 2.0])
T = np.full_like(P, 298.15)
q = 1.5 * 0.8 * P / (1 + 0.8 * P)

fit = fit_isotherm(P, q, T, "Langmuir", mode="single")

# parameters[1] = log10(b), so b = 10**parameters[1]
fit.parameters
```

## Mixture IAST

```python
components = [
    {"MoleculeName": "A", "isotherms": [["Langmuir", 1.0, 0.5]]},
    {"MoleculeName": "B", "isotherms": [["Langmuir", 2.0, 0.2]]},
]

parsed = parse(components)
iast([1.0, 0.5], parsed=parsed)
```

## Crank Diffusion Uptake

```python
times = np.linspace(0, 10, 6)
q_over_qinf = crank_uptake(times, D_R2=1e-3)

[(float(ti), float(qi)) for ti, qi in zip(times, q_over_qinf)]
```

## ZLC Model

```python
t, c, c_preload, q, q_preload, sigsum = linmodel(
    Lsat=20,
    DR2=6.4e-3,
    gamma=0.001,
    tsat=50,
    t_end=80,
    NN="auto",
)

t[:5], c[:5]
```
